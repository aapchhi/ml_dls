# Метрика

Перед решением любой задачи важно понимать, как будет оцениваться ваше решение. В данном случае мы используем стандартную для задачи классификации метрику ROC-AUC. Ее можно вычислить, используя только предсказанные вероятности и истинные классы без конкретного порога классификации + она раотает даже если классы в данных сильно несбалансированны (примеров одного класса в десятки раз больше примеров длугого). Именно поэтому она очень удобна для соревнований.

Посчитать ее легко:


```python
from sklearn.metrics import roc_auc_score

y_true = [
    0,
    1,
    1,
    0,
    1
]

y_predictions = [
    0.1,
    0.9,
    0.4,
    0.6,
    0.61
]

roc_auc_score(y_true, y_predictions)
```

```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
```

```python
!pip install catboost
```

```python
scaler = StandardScaler() #числовые переменные стандартизированы; категориальные --- в виде бинарных признаков
data[num_cols] = scaler.fit_transform(data[num_cols])
data_proc = pd.concat([data[num_cols], pd.get_dummies(data[cat_cols], drop_first=True)], axis=1) #данные для логистической регрессии
```

```python
data_proc
```

```python
X_train, X_test, y_train, y_test = train_test_split(data_proc, data[target_col], test_size=0.2)
```

```python
logreg = LogisticRegressionCV(
    Cs=[100, 10, 1, 0.1, 0.01, 0.001],  # тестируемые значения C
    cv=5,                                # 5-fold кросс-валидация
    scoring='roc_auc',
    random_state=42,
    max_iter=1000
) #логрег с кросс-валидацией; метрика ROC-AUC
logreg.fit(X_train, y_train)
logreg_pred = logreg.predict_proba(X_test)
```

```python
# После обучения
print("Лучший параметр C:", logreg.C_[0])
print("Лучший ROC-AUC на кросс-валидации:", logreg.scores_[1].mean(axis=0).max())

# ROC-AUC на тесте
test_auc = roc_auc_score(y_test, logreg_pred[:, 1])
print(f"ROC-AUC на тестовой выборке: {test_auc:.4f}")
```

## Применение градиентного бустинга (2 балла)

Если вы хотите получить баллы за точный ответ, то стоит попробовать градиентный бустинг. Часто градиентный бустинг с параметрами по умолчанию даст вам 80% результата за 0% усилий.

Мы будем использовать catboost, поэтому нам не надо кодировать категориальные признаки. Catboost сделает это сам (в .fit() надо передать cat_features=cat_cols). А численные признаки нормировать для моделей, основанных на деревьях не нужно.

1) Разделите выборку на train/valid. Протестируйте catboost cо стандартными параметрами.

2) Протестируйте разные занчения параметров количества деревьев и learning_rate'а и выберите лучшую по метрике ROC-AUC комбинацию.

(Дополнительно) Есть некоторые сложности с тем, чтобы использовать CatBoostClassifier вместе с GridSearchCV, поэтому мы не просим использовать кроссвалидацию. Но можете попробовать)

```python
from catboost import CatBoostClassifier
```

```python
# Разделение на train/valid
X_train, X_valid, y_train, y_valid = train_test_split(
    data.drop(columns=[target_col]),  # все фичи кроме целевой
    data[target_col],
    test_size=0.2,
    random_state=42,
    stratify=data[target_col]
)
```

```python
# CatBoost со стандартными параметрами
catboost_default = CatBoostClassifier(
    cat_features=cat_cols,  # указываем категориальные колонки
    random_state=42,
    n_estimators=100,
    learning_rate=0.05,
    verbose=False  # отключаем вывод обучения
)
```

```python
catboost_default.fit(X_train, y_train)
```

```python
# Предсказание и оценка
y_pred_proba_default = catboost_default.predict_proba(X_valid)[:, 1]
auc_default = roc_auc_score(y_valid, y_pred_proba_default)

print(f" CatBoost (стандартные параметры) ROC-AUC: {auc_default:.4f}")
```

```python
n_estimators_list = [100, 200, 500]
learning_rates = [0.01, 0.05, 0.1]

best_auc = 0
best_params = {}

print("n_estimators | learning_rate | ROC-AUC")
print("-" * 40)

for n_est in n_estimators_list:
    for lr in learning_rates:
        model = CatBoostClassifier(
            n_estimators=n_est,
            learning_rate=lr,
            cat_features=cat_cols,
            random_state=42,
            verbose=False
        )
        
        model.fit(X_train, y_train)
        
        # Оценка на valid
        y_pred_proba = model.predict_proba(X_valid)[:, 1]
        auc_score = roc_auc_score(y_valid, y_pred_proba)
        
        print(f"{n_est:12} | {lr:13} | {auc_score:.4f}")
        
        # Сохранение лучшей модели
        if auc_score > best_auc:
            best_auc = auc_score
            best_params = {'n_estimators': n_est, 'learning_rate': lr}
            best_model = model

print(f"\nЛучшие параметры: {best_params}")
print(f"Лучший ROC-AUC: {best_auc:.4f}")
```
